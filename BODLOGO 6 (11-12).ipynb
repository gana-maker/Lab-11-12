```python
#BODLOGO 35

# ХИЕХ олох функц
def hieh(a, b):
    while b != 0:
        a, b = b, a % b
    return abs(a)

# Бутархайн утгыг оруулах
a = int(input("Хүртвэр a = "))
b = int(input("Хуваарь b = "))

if b == 0:
    print("Алдаа: Хуваарь 0 байж болохгүй.")
else:
    # ХИЕХ олох
    d = hieh(a, b)

    # Бутархайг хураах
    a = a // d
    b = b // d

    if b < 0:
        a = -a
        b = -b

    print("Хураасан бутархай:", f"{a}/{b}")
```

```python
#BODLOGO 4
# Фибоначийн тоо шалгах функц
def fibonacci(n):
    a, b = 0, 1
    i = 1

    while a < n:
        a, b = b, a + b
        i += 1

    if a == n:
        print(n, "нь Фибоначийн тоо мөн.")
        print("Дарааллын", i, "-р гишүүн.")
    else:
        print(n, "нь Фибоначийн тоо биш.")

# Натурал тоо оруулах
n = int(input("Натурал тоо оруулна уу: "))

if n > 0:
    fibonacci(n)
else:
    print("Эерэг натурал тоо оруулна уу.")
```

```python
#BODLOGO 5
class Rectangle:
    def __init__(self, a=None, b=None, name=None,
                 filename=None, other=None):

        # 1. Гараас өгөгдөл авах
        if a is None and filename is None and other is None:
            self.name = input("Нэр: ")

            # Зөвхөн тоон утга оруулахыг шаардах
            while True:
                try:
                    self.a = float(input("Урт a = "))
                    break
                except ValueError:
                    print("Алдаа: Уртыг зөвхөн тоогоор оруулна уу!")

            while True:
                try:
                    self.b = float(input("Өргөн b = "))
                    break
                except ValueError:
                    print("Алдаа: Өргөнийг зөвхөн тоогоор оруулна уу!")

        # 2. Файлаас өгөгдөл авах
        elif filename is not None:
            with open(filename, "r", encoding="utf-8") as f:
                data = f.read().split()
                self.name = data[0]
                self.a = float(data[1])
                self.b = float(data[2])

        # 3. Өөр объектоос хуулах
        elif other is not None:
            self.name = other.name
            self.a = other.a
            self.b = other.b

        # 4. Параметраар өгөгдөл авах
        else:
            self.a = float(a)
            self.b = float(b)
            self.name = name

    # Талбай олох
    def area(self):
        return self.a * self.b

    # Периметр олох
    def perimeter(self):
        return 2 * (self.a + self.b)

    # Мэдээлэл хэвлэх
    def show(self):
        print("Нэр:", self.name)
        print("Урт:", self.a)
        print("Өргөн:", self.b)
        print("Талбай:", self.area())
        print("Периметр:", self.perimeter())
        print("------------------------")


# 1-р объект: Гараас унших
print("1. Гараас өгөгдөл оруулах")
r1 = Rectangle()

# 2-р объект: Параметраар үүсгэх
print("\n2. Параметраар үүсгэх")
r2 = Rectangle(10, 5, "Rectangle2")

# 3-р объект: Файлаас унших
with open("rectangle.txt", "w", encoding="utf-8") as f:
    f.write("Rectangle3 8 4")

print("\n3. Файлаас унших")
r3 = Rectangle(filename="rectangle.txt")

# 4-р объект: Өөр объектоос хуулах
print("\n4. Өөр объектоос хуулах")
r4 = Rectangle(other=r2)

# Бүх объектын үр дүнг хэвлэх
print("\n===== ҮР ДҮН =====")
r1.show()
r2.show()
r3.show()
r4.show()
```

```python
#bodlogo 6

import math

class EqQuad:
    def __init__(self, a=None, b=None, c=None,
                 filename=None, other=None):

        # 1. Гараас өгөгдөл авах
        if a is None and filename is None and other is None:
            self.a = float(input("a = "))
            self.b = float(input("b = "))
            self.c = float(input("c = "))

        # 2. Файлаас өгөгдөл авах
        elif filename is not None:
            with open(filename, "r", encoding="utf-8") as f:
                data = f.read().split()
                self.a = float(data[0])
                self.b = float(data[1])
                self.c = float(data[2])

        # 3. Өөр объектоос хуулах
        elif other is not None:
            self.a = other.a
            self.b = other.b
            self.c = other.c

        # 4. Параметраар өгөгдөл авах
        else:
            self.a = float(a)
            self.b = float(b)
            self.c = float(c)

    # Квадрат тэгшитгэлийн бодолт
    def bodolt(self):
        a, b, c = self.a, self.b, self.c

        print(f"Тэгшитгэл: {a}x² + ({b})x + ({c}) = 0")

        if a == 0:
            print("Квадрат тэгшитгэл биш.")
            if b != 0:
                print("x =", -c / b)
            elif c == 0:
                print("Хязгааргүй олон шийдтэй.")
            else:
                print("Шийдгүй.")
            return

        D = b**2 - 4*a*c
        print("Дискриминант D =", D)

        if D > 0:
            x1 = (-b + math.sqrt(D)) / (2*a)
            x2 = (-b - math.sqrt(D)) / (2*a)
            print("Хоёр бодит шийдтэй:")
            print("x1 =", x1)
            print("x2 =", x2)

        elif D == 0:
            x = -b / (2*a)
            print("Нэг давхар бодит шийдтэй:")
            print("x =", x)

        else:
            print("Бодит шийдгүй.")


# 1-р объект: Гараас утга авах
print("1. Гараас өгөгдөл оруулах")
eq1 = EqQuad()

# 2-р объект: Параметраар үүсгэх
print("\n2. Параметраар үүсгэх")
eq2 = EqQuad(1, -5, 6)

# 3-р объект: Файлаас унших
with open("equation.txt", "w", encoding="utf-8") as f:
    f.write("1 2 1")

print("\n3. Файлаас унших")
eq3 = EqQuad(filename="equation.txt")

# 4-р объект: Өөр объектоос хуулах
print("\n4. Өөр объектоос хуулах")
eq4 = EqQuad(other=eq2)

# Бүх объектын бодолт
print("\n===== ҮР ДҮН =====")

print("\n1-р объект")
eq1.bodolt()

print("\n2-р объект")
eq2.bodolt()

print("\n3-р объект")
eq3.bodolt()

print("\n4-р объект")
eq4.bodolt()
```
